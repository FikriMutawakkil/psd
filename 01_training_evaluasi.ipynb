# Evaluasi Model Klasifikasi LULC Jawa Timur

Notebook ini mendokumentasikan proses pelatihan dan evaluasi model
klasifikasi tutupan lahan di Jawa Timur menggunakan citra Sentinel-2.

## Tujuan

- Mendokumentasikan proses klasifikasi tutupan lahan.
- Menjelaskan evaluasi performa model.
- Mendokumentasikan hasil klasifikasi.

## Catatan

Notebook ini merupakan dokumentasi awal. Hasil evaluasi dan metrik
model perlu diisi berdasarkan hasil eksperimen yang sebenarnya.
