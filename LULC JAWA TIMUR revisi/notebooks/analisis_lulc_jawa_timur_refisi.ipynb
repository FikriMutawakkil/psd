
# Klasifikasi Land Use/Land Cover (LULC) Jawa Timur

## 1. Business Understanding

### 1.1 Latar Belakang
Pemanfaatan citra satelit Sentinel-2 dan Machine Learning digunakan untuk mengidentifikasi serta memetakan tutupan lahan di wilayah Jawa Timur.

### 1.2 Tujuan

### 1.3 Fokus Penelitian
Mengevaluasi penggunaan band Sentinel-2 B4, B8A, dan B11 untuk membedakan hutan mangrove dengan hutan biasa.

### 1.4 Hasil yang Diharapkan

### Cell 1 — Memuat dataset fitur
Membaca CSV hasil ekstraksi fitur Sentinel-2 dan menampilkan ukuran, kolom, serta contoh data. Jalankan sebelum analisis berikutnya.

```python
import pandas as pd
from pathlib import Path

# Lokasi dataset
CSV_PATH = Path("../data/sentinel_bands/fitur_lulc_jatim.csv")

# Membaca dataset
df = pd.read_csv(CSV_PATH)

# Menampilkan ukuran dataset
print("Jumlah baris dan kolom:", df.shape)

# Menampilkan nama kolom
print("\nNama kolom:")
print(df.columns.tolist())

# Menampilkan lima baris pertama
display(df.head())
```

### Cell 2 — Pemeriksaan kualitas data
Memeriksa nilai kosong, duplikasi, distribusi label, dan statistik fitur. Gunakan hasil ini untuk menilai apakah data perlu dibersihkan.

```python
# 1. Memeriksa nilai kosong
print("=== NILAI KOSONG ===")
print(df.isnull().sum())

# 2. Memeriksa data duplikat
print("\n=== DATA DUPLIKAT ===")
print("Jumlah baris duplikat:", df.duplicated().sum())

# 3. Memeriksa distribusi kelas
print("\n=== DISTRIBUSI KELAS ===")
print(df["class"].value_counts())

# 4. Memeriksa statistik fitur numerik
print("\n=== STATISTIK FITUR ===")
display(df.describe().round(4))
```

### Cell 3 — Pemeriksaan redundansi indeks
Mengecek hubungan NDBI dan NDMI. Pada perhitungan yang digunakan sebelumnya, keduanya dapat saling berlawanan; karena itu model utama di bawah mempertahankan NDMI dan tidak menggunakan NDBI.

```python

# Memeriksa hubungan NDBI dan NDMI
print("Korelasi NDBI dengan NDMI:")
print(df[["NDBI", "NDMI"]].corr().round(4))

# Memeriksa apakah keduanya saling berlawanan
selisih = (df["NDBI"] + df["NDMI"]).abs()

print("\nNilai maksimum |NDBI + NDMI|:")
print(selisih.max())

# Memeriksa nama label kelas
print("\nLabel kelas:")
print(sorted(df["class"].unique()))
```

### Cell 4 — Standardisasi label
Menyeragamkan salah ketik `fram land` menjadi `farm land` tanpa mengubah dataset asli `df`.

```python

# Membuat salinan data agar data asli tetap aman
data = df.copy()

# Menyeragamkan nama kelas lahan pertanian
data["class"] = data["class"].replace({
    "fram land": "farm land"
})

# Memeriksa label kelas setelah perbaikan
print("Label kelas setelah standardisasi:")
print(sorted(data["class"].unique()))

print("\nDistribusi kelas:")
print(data["class"].value_counts())
```

### Cell 5 — Menentukan target dan fitur
Menetapkan `class` sebagai target dan menyusun kolom fitur spektral/indeks yang digunakan untuk pemodelan.

```python

# Menentukan kolom target
target = "class"

# Menentukan fitur spektral dan indeks
fitur_lengkap = [
    "B2", "B3", "B4", "B5", "B6",
    "B7", "B8", "B8A", "B11", "B12",
    "NDVI", "NDWI", "MNDWI", "NDMI"
]

# Memisahkan fitur dan target
X = data[fitur_lengkap].copy()
y = data[target].copy()

# Memeriksa hasil pemisahan
print("Ukuran fitur (X):", X.shape)
print("Ukuran target (y):", y.shape)

print("\nContoh fitur:")
display(X.head())

print("\nContoh target:")
display(y.head())
```

### Cell 6 — Menyiapkan dua set fitur
Model utama memakai 14 fitur (NDBI dikeluarkan dan NDMI dipertahankan). Model pembanding hanya memakai B4, B8A, dan B11 sesuai fokus pembeda mangrove. Cell ini juga memeriksa nilai tak hingga.

```python
import numpy as np


# Model utama menggunakan 14 fitur; NDBI dihapus karena redundan dengan NDMI
fitur_model = [
    "B2", "B3", "B4", "B5", "B6",
    "B7", "B8", "B8A", "B11", "B12",
    "NDVI", "NDWI", "MNDWI", "NDMI"
]

X_lengkap = data[fitur_model].copy()

# Fitur khusus sesuai arahan dosen
fitur_mangrove = ["B4", "B8A", "B11"]
X_mangrove = data[fitur_mangrove].copy()

print("Jumlah fitur model lengkap:", len(fitur_model))
print("Kolom model lengkap:", fitur_model)

print("\nJumlah fitur model mangrove:", len(fitur_mangrove))
print("Kolom model mangrove:", fitur_mangrove)

print("\nPemeriksaan nilai tak hingga:")
print("Model lengkap:", np.isinf(X_lengkap.to_numpy()).sum())
print("Model mangrove:", np.isinf(X_mangrove.to_numpy()).sum())
```

### Cell 7 — Membagi data latih dan uji
Membagi data model utama menjadi 75% data latih dan 25% data uji dengan stratifikasi agar proporsi kelas relatif terjaga.

```python

from sklearn.model_selection import train_test_split

# Membagi data: 75% latih dan 25% uji
X_train, X_test, y_train, y_test = train_test_split(
    X_lengkap,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Memeriksa ukuran hasil pembagian
print("Jumlah data awal :", len(data))
print("Data latih       :", len(X_train))
print("Data uji         :", len(X_test))

print("\nDistribusi kelas data latih:")
print(y_train.value_counts())

print("\nDistribusi kelas data uji:")
print(y_test.value_counts())

```

### Cell 8 — Melatih Random Forest utama
Melatih Random Forest dengan 300 pohon pada fitur model utama.

```python

from sklearn.ensemble import RandomForestClassifier

# Membuat model Random Forest
rf_lengkap = RandomForestClassifier(
    n_estimators=300,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Melatih model menggunakan data latih
rf_lengkap.fit(X_train, y_train)

print("Model Random Forest berhasil dilatih.")
print("Jumlah pohon:", rf_lengkap.n_estimators)
print("Jumlah fitur:", rf_lengkap.n_features_in_)
print("Jumlah kelas:", len(rf_lengkap.classes_))
print("Kelas model:", rf_lengkap.classes_)
```

### Cell 9 — Evaluasi model utama
Menghitung akurasi, classification report, dan confusion matrix pada data uji.

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Prediksi data uji
y_pred = rf_lengkap.predict(X_test)

# Menghitung akurasi
akurasi = accuracy_score(y_test, y_pred)
print(f"Akurasi model: {akurasi:.4f} ({akurasi * 100:.2f}%)")

# Laporan evaluasi setiap kelas
print("\nLaporan klasifikasi:")
print(classification_report(y_test, y_pred, zero_division=0))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred, labels=rf_lengkap.classes_)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=rf_lengkap.classes_
)

fig, ax = plt.subplots(figsize=(9, 7))
disp.plot(ax=ax, cmap="Blues", xticks_rotation=45, values_format="d")
plt.title("Confusion Matrix - Random Forest 14 Fitur")
plt.tight_layout()
plt.show()
```

### Cell 10 — Melatih model pembanding B4, B8A, B11
Menggunakan pembagian data yang sama untuk membandingkan performa tiga band yang ditekankan pada fokus mangrove.

```python

# Membagi data dengan fitur khusus mangrove
X_train_mangrove, X_test_mangrove, y_train_mangrove, y_test_mangrove = train_test_split(
    X_mangrove,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Membuat model Random Forest khusus B4, B8A, B11
rf_mangrove = RandomForestClassifier(
    n_estimators=300,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Melatih model
rf_mangrove.fit(X_train_mangrove, y_train_mangrove)

# Prediksi data uji
y_pred_mangrove = rf_mangrove.predict(X_test_mangrove)

# Evaluasi
akurasi_mangrove = accuracy_score(y_test_mangrove, y_pred_mangrove)

print(f"Akurasi model B4-B8A-B11: {akurasi_mangrove:.4f} ({akurasi_mangrove * 100:.2f}%)")

print("\nLaporan klasifikasi:")
print(classification_report(
    y_test_mangrove,
    y_pred_mangrove,
    zero_division=0
))
```

### Cell 11 — Membandingkan model
Merangkum Accuracy, Macro F1, dan F1 kelas mangrove untuk model utama dan model tiga band. Angka ini berasal dari pembagian acak; sampel berdekatan secara spasial dapat membuat evaluasi terlalu optimistis.

### Cell 20 — Membandingkan label GeoJSON dengan raster
Menghitung kelas raster dominan pada setiap poligon sampel. Ini pemeriksaan kecocokan visual/label, bukan evaluasi akurasi independen.

```python

from rasterio.mask import mask
from collections import Counter

# Memastikan CRS GeoJSON sama dengan raster
with rasterio.open(raster_path) as src:
    if gdf.crs != src.crs:
        gdf_raster = gdf.to_crs(src.crs)
    else:
        gdf_raster = gdf.copy()

    hasil_cocok = []

    for _, row in gdf_raster.iterrows():
        try:
            out_image, _ = mask(
                src,
                [row.geometry],
                crop=True,
                filled=False
            )

            values = out_image[0].compressed()
            values = values[(values != 255)]

            if len(values) > 0:
                counts = Counter(values.tolist())
                kelas_dominan = counts.most_common(1)[0][0]
                hasil_cocok.append({
                    "class_geojson": row["class"],
                    "nilai_raster_dominan": kelas_dominan
                })

        except Exception:
            continue

hasil_cocok_df = pd.DataFrame(hasil_cocok)

print("Jumlah poligon yang berhasil dicocokkan:", len(hasil_cocok_df))
display(
    pd.crosstab(
        hasil_cocok_df["class_geojson"],
        hasil_cocok_df["nilai_raster_dominan"]
    )
)
```

### Cell 23 — Peta interaktif sampel GeoJSON
Membuat peta satelit dengan checkbox untuk mengaktifkan atau menonaktifkan poligon digitasi per kelas.

```python

import folium
from IPython.display import display

# Titik tengah wilayah digitasi
minx, miny, maxx, maxy = gdf.total_bounds
titik_tengah = [(miny + maxy) / 2, (minx + maxx) / 2]

# Membuat peta dengan basemap satelit
peta = folium.Map(
    location=titik_tengah,
    zoom_start=8,
    tiles=None
)

# Menambahkan basemap satelit Esri
folium.TileLayer(
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Imagery/MapServer/tile/{z}/{y}/{x}",
    attr="Esri, Maxar, Earthstar Geographics",
    name="Citra Satelit",
    overlay=False,
    control=True
).add_to(peta)

# Warna setiap kelas GeoJSON
warna_geojson = {
    "build up": "#D94F4F",
    "fram land": "#E6C84F",
    "water body": "#3685D6",
    "forest": "#27834A",
    "hutan mangrove": "#35B8B0"
}

# Membuat satu layer terpisah untuk setiap kelas
for kelas, warna in warna_geojson.items():
    data_kelas = gdf[gdf["class"] == kelas]

    if not data_kelas.empty:
        layer = folium.FeatureGroup(
            name=kelas.title(),
            show=True
        )

        folium.GeoJson(
            data_kelas.to_json(),
            style_function=lambda feature, warna=warna: {
                "fillColor": warna,
                "color": warna,
                "weight": 1,
                "fillOpacity": 0.55
            },
            tooltip=folium.GeoJsonTooltip(
                fields=["class"],
                aliases=["Kelas:"]
            )
        ).add_to(layer)

        layer.add_to(peta)

# Menambahkan kontrol layer
folium.LayerControl(collapsed=False).add_to(peta)

# Menyesuaikan tampilan dengan batas GeoJSON
peta.fit_bounds([[miny, minx], [maxy, maxx]])

display(peta)
```

### Cell 25 — Peta interaktif raster dan sampel
Membuat peta Folium dari raster yang diperkecil untuk pratinjau. Setiap kelas raster memiliki checkbox sendiri; sampel GeoJSON ditambahkan sebagai layer terpisah dan secara default disembunyikan. File HTML disimpan di `data/processed`.

```python

# ============================================================
# CELL 25 — PETA INTERAKTIF RASTER LULC DAN SAMPEL GEOJSON
# Output langsung di notebook VS Code, tanpa membuat HTML baru
# ============================================================

import numpy as np
import rasterio
import geopandas as gpd
import folium

from pathlib import Path
from rasterio.enums import Resampling
from PIL import Image
from io import BytesIO
import base64
from IPython.display import display


# 1. Menentukan lokasi file


from pathlib import Path

PROJECT_DIR = Path(
    r"D:\semester 5\PROYEK SAINS DATA\LULC_JAWA_TIMUR\LULC JAWA TIMUR revisi"
)

RASTER_PATH = (
    PROJECT_DIR / "data" / "lulc_jawa_timur_rf_revisi.tif"
)

GEOJSON_PATH = (
    PROJECT_DIR / "data" / "geojson" / "jawatimur.geojson"
)

print("Raster ditemukan:", RASTER_PATH.exists())
print("Lokasi raster:", RASTER_PATH)

print("GeoJSON ditemukan:", GEOJSON_PATH.exists())
print("Lokasi GeoJSON:", GEOJSON_PATH)

if not RASTER_PATH.exists():
    raise FileNotFoundError(f"Raster tidak ditemukan: {RASTER_PATH}")

if not GEOJSON_PATH.exists():
    raise FileNotFoundError(f"GeoJSON tidak ditemukan: {GEOJSON_PATH}")


# 2. Membaca raster dalam resolusi kecil agar hemat memori
with rasterio.open(RASTER_PATH) as src:
    factor = min(1, 1400 / max(src.width, src.height))
    height = max(1, int(src.height * factor))
    width = max(1, int(src.width * factor))

    raster = src.read(
        1,
        out_shape=(height, width),
        resampling=Resampling.nearest
    )

    bounds = src.bounds
    nodata = src.nodata

raster_bounds = [
    [bounds.bottom, bounds.left],
    [bounds.top, bounds.right]
]


# 3. Membaca sampel GeoJSON
gdf = gpd.read_file(GEOJSON_PATH)

if gdf.crs is not None and gdf.crs.to_epsg() != 4326:
    gdf = gdf.to_crs(epsg=4326)

# Menyamakan label kelas
gdf["class"] = (
    gdf["class"]
    .astype(str)
    .str.strip()
    .str.lower()
    .replace({"fram land": "farm land"})
)

kelas = {
    0: ("Farm Land", "farm land", "#F1D34D"),
    1: ("Build Up", "build up", "#D73027"),
    2: ("Water Body", "water body", "#2878B5"),
    3: ("Forest", "forest", "#238443"),
    4: ("Hutan Mangrove", "hutan mangrove", "#20B2AA")
}


# 4. Membuat peta interaktif langsung di notebook
pusat = [
    (bounds.bottom + bounds.top) / 2,
    (bounds.left + bounds.right) / 2
]

peta = folium.Map(
    location=pusat,
    zoom_start=8,
    tiles=None,
    control_scale=True
)

folium.TileLayer(
    "OpenStreetMap",
    name="OpenStreetMap",
    overlay=False
).add_to(peta)

folium.TileLayer(
    tiles=(
        "https://server.arcgisonline.com/ArcGIS/rest/services/"
        "World_Imagery/MapServer/tile/{z}/{y}/{x}"
    ),
    attr="Tiles © Esri",
    name="Citra Satelit",
    overlay=False
).add_to(peta)


# 5. Menampilkan raster per kelas sebagai layer terpisah
for class_id, (nama, label, warna) in kelas.items():
    rgba = np.zeros((height, width, 4), dtype=np.uint8)

    rgb = tuple(
        int(warna[i:i + 2], 16)
        for i in (1, 3, 5)
    )

    mask = raster == class_id

    rgba[mask, 0] = rgb[0]
    rgba[mask, 1] = rgb[1]
    rgba[mask, 2] = rgb[2]
    rgba[mask, 3] = 170

    if nodata is not None:
        rgba[raster == nodata, 3] = 0

    buffer = BytesIO()
    Image.fromarray(rgba).save(buffer, format="PNG")

    encoded = base64.b64encode(
        buffer.getvalue()
    ).decode("utf-8")

    folium.raster_layers.ImageOverlay(
        image="data:image/png;base64," + encoded,
        bounds=raster_bounds,
        name=f"Raster — {nama}",
        opacity=0.85,
        interactive=False,
        show=True
    ).add_to(peta)


# 6. Menampilkan sampel GeoJSON per kelas
for class_id, (nama, label, warna) in kelas.items():
    sampel_kelas = gdf[gdf["class"] == label].copy()

    if sampel_kelas.empty:
        continue

    feature_group = folium.FeatureGroup(
        name=f"Sampel — {nama}",
        show=True
    )

    folium.GeoJson(
        data=sampel_kelas.__geo_interface__,
        style_function=lambda feature, color=warna: {
            "color": color,
            "fillColor": color,
            "weight": 2,
            "fillOpacity": 0.25
        },
        tooltip=folium.GeoJsonTooltip(
            fields=["class"],
            aliases=["Kelas:"]
        )
    ).add_to(feature_group)

    feature_group.add_to(peta)


# 7. Menambahkan kontrol layer
folium.LayerControl(
    collapsed=False
).add_to(peta)

# Atur tampilan awal mencakup seluruh raster
peta.fit_bounds(raster_bounds)

# 8. Tampilkan langsung di output notebook
# Tidak menggunakan peta.save(), jadi tidak membuat HTML baru.
display(peta)

print("Jumlah sampel GeoJSON:", len(gdf))
print("Kolom GeoJSON:", list(gdf.columns))
print("Peta raster dan sampel berhasil ditampilkan.")
```
